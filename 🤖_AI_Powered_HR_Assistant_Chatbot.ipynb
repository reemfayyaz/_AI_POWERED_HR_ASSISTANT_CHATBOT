#🤖 AI-Powered HR Assistant Chatbot

#Project Overview

SmartHR AI is an intelligent NLP-based HR chatbot designed to provide quick and automated answers to common employee-related queries. The chatbot can understand user questions related to leave, attendance, salary, payroll, work-from-home, benefits, recruitment, onboarding, resignation, and HR policies.

The system uses TF-IDF for text feature extraction and Logistic Regression for intent classification. Based on the detected intent, the chatbot provides an appropriate HR response. The trained chatbot model is serialized into a .pkl file using Joblib and integrated with a Streamlit web application.

# Objectives
Automate repetitive HR queries.
Provide instant responses to employees.
Reduce the workload of HR departments.
Implement NLP-based intent classification.
Build an interactive chatbot interface.
Deploy the chatbot as a web application.
Create a reusable .pkl machine-learning model.

```python
# ============================================
# TRAIN HR CHATBOT MODEL
# ============================================

# ============================================
# IMPORT REQUIRED MACHINE LEARNING LIBRARIES
# ============================================

# TfidfVectorizer converts text data into numerical
# features that a Machine Learning model can understand.
#
# TF-IDF stands for:
# Term Frequency - Inverse Document Frequency.
#
# It gives higher importance to words that are
# meaningful for a particular document or sentence.

from sklearn.feature_extraction.text import TfidfVectorizer


# Logistic Regression is a classification algorithm.
#
# In our HR Chatbot, it is used to classify an
# employee's question into the correct HR intent,
# such as:
# - leave
# - salary
# - attendance
# - work_from_home
# - payroll
# - benefits
# - recruitment
# etc.

from sklearn.linear_model import LogisticRegression


# Pipeline allows us to combine multiple Machine
# Learning steps into a single workflow.
#
# Here, the pipeline will perform:
#
# Text → TF-IDF Transformation → Classification
#
# This makes the complete NLP workflow easier
# to train and use later.

from sklearn.pipeline import Pipeline


# train_test_split divides the dataset into two parts:
#
# Training Data → Used to train the chatbot.
# Testing Data  → Used to evaluate the chatbot.
#
# This helps us check whether the model can
# correctly classify new/unseen questions.

from sklearn.model_selection import train_test_split


# accuracy_score calculates the percentage of
# predictions that were classified correctly.
#
# classification_report provides detailed
# performance metrics such as:
# - Precision
# - Recall
# - F1-score
# - Support

from sklearn.metrics import accuracy_score, classification_report


# ============================================
# DEFINE INPUT FEATURES AND TARGET LABEL
# ============================================

# "text" contains employee questions.
#
# Example:
# "How many days of leave can I take?"
# "Can I work from home?"
# "When will I receive my salary?"

# These questions are the INPUT features
# that the chatbot will learn from.

X = df["text"]


# "intent" contains the category/label associated
# with each employee question.
#
# Example:
# leave
# salary
# attendance
# payroll
# benefits
#
# This is the TARGET variable that the model
# needs to predict.

y = df["intent"]


# ============================================
# TRAIN / TEST SPLIT
# ============================================

# Split the dataset into training and testing sets.
#
# Training data is used to teach the model.
# Testing data is used to evaluate the model
# on previously unseen examples.

X_train, X_test, y_train, y_test = train_test_split(

    # Input text used for training and testing.
    X,

    # Corresponding intent labels.
    y,

    # 20% of the dataset will be used for testing.
    #
    # 80% → Training Data
    # 20% → Testing Data
    test_size=0.20,

    # random_state ensures that the same split
    # is generated every time the code runs.
    #
    # This makes our experiment reproducible.
    random_state=42,

    # stratify=y keeps the proportion of each intent
    # approximately the same in both training and
    # testing datasets.
    #
    # This is especially important for classification
    # problems with multiple intent categories.
    stratify=y
)


# ============================================
# CREATE NLP MACHINE LEARNING PIPELINE
# ============================================

# Pipeline combines text preprocessing and
# Machine Learning classification into one object.
#
# The complete flow is:
#
# Employee Question
#        ↓
# TF-IDF Vectorization
#        ↓
# Numerical Features
#        ↓
# Logistic Regression
#        ↓
# Predicted HR Intent

model = Pipeline([

    # ----------------------------------------
    # STEP 1: TF-IDF TEXT VECTORIZATION
    # ----------------------------------------

    (
        "tfidf",

        TfidfVectorizer(

            # Convert all text into lowercase.
            #
            # Example:
            # "Salary" → "salary"
            #
            # This prevents the model from treating
            # uppercase and lowercase versions as
            # completely different words.
            lowercase=True,

            # ngram_range=(1, 2) means the vectorizer
            # considers both:
            #
            # 1-word combinations → Unigrams
            # 2-word combinations → Bigrams
            #
            # Example:
            # "salary"       → unigram
            # "salary slip" → bigram
            #
            # Bigrams help the chatbot understand
            # meaningful word combinations.
            ngram_range=(1, 2),

            # sublinear_tf=True applies logarithmic
            # scaling to term frequency.
            #
            # This can reduce the influence of words
            # that appear many times in a document and
            # can make text classification more robust.
            sublinear_tf=True
        )
    ),

    # ----------------------------------------
    # STEP 2: LOGISTIC REGRESSION CLASSIFIER
    # ----------------------------------------

    (
        "classifier",

        LogisticRegression(

            # Increase the maximum number of iterations
            # allowed during model optimization.
            #
            # max_iter=2000 gives the algorithm enough
            # iterations to converge on our multi-class
            # text classification problem.
            max_iter=2000,

            # C controls the regularization strength.
            #
            # Higher C generally means weaker regularization
            # and allows the model to fit the training data
            # more closely.
            #
            # Here we use C=5.
            C=5
        )
    )
])


# ============================================
# TRAIN THE HR CHATBOT MODEL
# ============================================

# model.fit() trains the complete pipeline.
#
# First:
# TF-IDF converts the training text into numerical
# feature vectors.
#
# Then:
# Logistic Regression learns the relationship
# between those numerical features and HR intents.
#
# After this step, the model has learned how to
# classify employee questions into HR categories.

model.fit(X_train, y_train)


# ============================================
# MAKE PREDICTIONS ON TEST DATA
# ============================================

# model.predict() uses the trained pipeline to
# predict the HR intent for each unseen question
# in the testing dataset.
#
# The prediction process automatically performs:
#
# Test Text
#     ↓
# TF-IDF Transformation
#     ↓
# Logistic Regression
#     ↓
# Predicted Intent

y_pred = model.predict(X_test)


# ============================================
# CALCULATE MODEL ACCURACY
# ============================================

# Compare the actual intent labels (y_test)
# with the predicted labels (y_pred).
#
# accuracy_score returns a value between 0 and 1.
#
# Example:
# 0.90 = 90% accuracy
# 0.95 = 95% accuracy

accuracy = accuracy_score(y_test, y_pred)


# ============================================
# DISPLAY MODEL TRAINING RESULT
# ============================================

# Print a visual separator to make the
# Google Colab output easier to read.

print("================================")


# Display a confirmation message indicating
# that the HR Chatbot model has been trained.

print("HR CHATBOT MODEL TRAINED")


# Print another separator for better readability.

print("================================")


# Convert accuracy from decimal format to percentage.
#
# Example:
# 0.9235 → 92.35%
#
# round(..., 2) keeps the result up to
# two decimal places.

print("Accuracy:", round(accuracy * 100, 2), "%")


# ============================================
# DISPLAY CLASSIFICATION REPORT
# ============================================

# classification_report provides detailed
# performance information for every HR intent.
#
# It includes:
#
# Precision:
# How many predicted examples were actually correct.
#
# Recall:
# How many actual examples of an intent were
# correctly identified.
#
# F1-score:
# Harmonic mean of precision and recall.
#
# Support:
# Number of actual test samples belonging
# to each intent.

print("\nClassification Report:")


# Display the complete classification report
# comparing actual and predicted HR intents.

print(classification_report(y_test, y_pred))
```

```python
# ============================================
# TEST HR CHATBOT
# ============================================

# This function takes the user's question as input,
# predicts the most relevant HR intent, checks the
# model's confidence, and returns the appropriate
# response from the HR knowledge base.

def get_hr_response(user_input, threshold=0.35):

    # ========================================
    # CLEAN USER INPUT
    # ========================================

    # strip() removes unnecessary spaces from
    # the beginning and end of the user's input.
    #
    # Example:
    # "   How can I apply for leave?   "
    #
    # becomes:
    # "How can I apply for leave?"

    user_input = user_input.strip()


    # ========================================
    # HANDLE EMPTY INPUT
    # ========================================

    # If the user submits an empty message,
    # there is nothing for the NLP model to classify.
    #
    # In that case, return a friendly message
    # instead of sending empty text to the model.

    if not user_input:

        return "Please enter a question."


    # ========================================
    # PREDICT INTENT PROBABILITIES
    # ========================================

    # predict_proba() returns the probability
    # associated with every possible HR intent.
    #
    # Example:
    #
    # leave            → 0.08
    # attendance       → 0.05
    # salary           → 0.12
    # work_from_home   → 0.71
    # payroll          → 0.04
    #
    # The probabilities indicate how confident
    # the model is about each possible intent.

    probabilities = model.predict_proba([user_input])[0]


    # ========================================
    # FIND HIGHEST PROBABILITY
    # ========================================

    # max() returns the highest probability
    # among all predicted intents.
    #
    # This represents the model's confidence
    # in its best prediction.

    max_probability = probabilities.max()


    # ========================================
    # IDENTIFY PREDICTED INTENT
    # ========================================

    # argmax() returns the position/index of
    # the highest probability.
    #
    # model.classes_ contains the corresponding
    # intent names in the same order.
    #
    # Therefore, we use the index of the highest
    # probability to identify the predicted intent.

    predicted_intent = model.classes_[probabilities.argmax()]


    # ========================================
    # CONFIDENCE THRESHOLD CHECK
    # ========================================

    # The threshold determines how confident
    # the chatbot must be before giving an answer.
    #
    # Default threshold = 0.35
    #
    # If the highest probability is below 0.35,
    # the model is not confident enough.
    #
    # In that situation, the chatbot should NOT
    # provide a potentially incorrect HR response.

    if max_probability < threshold:

        return (
            "I'm sorry, I couldn't confidently understand your HR question. "
            "Please try asking about leave, attendance, salary, payroll, "
            "benefits, recruitment, onboarding, resignation, or HR policies."
        )


    # ========================================
    # RETURN KNOWLEDGE BASE RESPONSE
    # ========================================

    # If the model's confidence is above the
    # required threshold, use the predicted intent
    # to retrieve the appropriate response from
    # our HR knowledge base.
    #
    # Example:
    #
    # predicted_intent = "leave"
    #
    # responses["leave"]
    #
    # returns the predefined HR leave response.

    return responses[predicted_intent]


# ============================================
# INTERACTIVE HR CHATBOT
# ============================================

# Display a clear heading before starting
# the interactive chatbot session.

print("====================================")


# Display the chatbot name.

print("       HR CHATBOT")


# Inform the user that typing "exit" will
# terminate the chatbot session.

print("Type 'exit' to stop")


# Display another separator for readability.

print("====================================")


# ============================================
# START CONTINUOUS CHAT LOOP
# ============================================

# while True creates a continuous conversation loop.
#
# The chatbot will continue accepting questions
# until the user enters one of the exit commands.

while True:


    # ========================================
    # GET USER INPUT
    # ========================================

    # input() pauses the program and waits for
    # the user to type an HR-related question.
    #
    # The question is stored in user_input.

    user_input = input("\nYou: ")


    # ========================================
    # CHECK EXIT COMMAND
    # ========================================

    # lower() converts the user's input to lowercase
    # so that different capitalization does not matter.
    #
    # Example:
    # "EXIT"
    # "Exit"
    # "exit"
    #
    # All become:
    # "exit"
    #
    # The chatbot will stop if the user enters:
    # exit, quit, or bye.

    if user_input.lower() in ["exit", "quit", "bye"]:


        # Display a goodbye message before
        # terminating the chatbot.

        print("HR Bot: Goodbye! 👋")


        # break immediately stops the while loop.

        break


    # ========================================
    # GENERATE HR RESPONSE
    # ========================================

    # Send the user's question to our
    # get_hr_response() function.
    #
    # The function:
    # 1. Cleans the input
    # 2. Predicts intent
    # 3. Calculates confidence
    # 4. Checks the threshold
    # 5. Retrieves the correct HR response

    answer = get_hr_response(user_input)


    # ========================================
    # DISPLAY CHATBOT RESPONSE
    # ========================================

    # Print the final response returned
    # by the HR chatbot.

    print("\nHR Bot:", answer)
```

```python
# ============================================
# SAVE HR CHATBOT AS PKL FILE
# ============================================

# Import the joblib library.
#
# joblib is commonly used in Machine Learning projects
# to save trained models and Python objects to a file.
#
# Here, we will use it to save our complete HR Chatbot
# package into a ".pkl" file.

import joblib


# Import the os module.
#
# The os module provides operating-system-related
# functionality.
#
# We will use it to check the size of the generated
# PKL file.

import os


# ============================================
# CREATE CHATBOT PACKAGE
# ============================================

# Create a Python dictionary containing all the
# components required to run our HR Chatbot.
#
# Instead of saving only the Machine Learning model,
# we are saving:
#
# 1. Trained NLP model
# 2. HR knowledge base responses
# 3. Confidence threshold
# 4. Chatbot name
#
# This allows us to load the complete chatbot later
# from a single PKL file.

chatbot_package = {

    # Store the trained TF-IDF + Logistic Regression
    # pipeline.
    #
    # This model is responsible for predicting
    # the HR intent from the user's question.

    "model": model,


    # Store the HR knowledge base.
    #
    # After predicting an intent, the application
    # will use this dictionary to retrieve the
    # appropriate HR response.

    "responses": responses,


    # Store the confidence threshold used by our
    # chatbot.
    #
    # If the model's highest prediction probability
    # is below 0.35, the chatbot will return a
    # fallback message instead of giving a potentially
    # incorrect HR response.

    "threshold": 0.35,


    # Store the name of our chatbot.
    #
    # This can later be displayed in the Streamlit
    # application interface.

    "name": "Professional HR Assistant"
}


# ============================================
# DEFINE PKL FILE PATH
# ============================================

# Define the location where the PKL file will be saved.
#
# "/content/" is the main working directory in
# Google Colab.
#
# The final file will be:
#
# /content/hr_chatbot.pkl

pkl_path = "/content/hr_chatbot.pkl"


# ============================================
# SAVE CHATBOT PACKAGE
# ============================================

# joblib.dump() serializes the complete chatbot
# package and saves it to the specified file.
#
# This means we do NOT need to retrain the model
# every time we want to use the chatbot.
#
# Later, we can simply load this PKL file using:
#
# joblib.load("/content/hr_chatbot.pkl")
#
# and retrieve the trained model, responses,
# threshold, and chatbot name.

joblib.dump(chatbot_package, pkl_path)


# ============================================
# DISPLAY SUCCESS MESSAGE
# ============================================

# Print a separator to make the output
# easier to read.

print("====================================")


# Confirm that the PKL file has been created.

print("PKL FILE CREATED SUCCESSFULLY")


# Print another separator.

print("====================================")


# ============================================
# DISPLAY FILE LOCATION
# ============================================

# Display the exact location of the generated
# PKL file.
#
# This helps us verify where the file has been
# saved in Google Colab.

print("File:", pkl_path)


# ============================================
# CHECK FILE SIZE
# ============================================

# os.path.getsize() returns the file size in bytes.
#
# We divide by 1024 to convert bytes into KB.
#
# round(..., 2) rounds the result to two decimal places.
#
# Example:
# 15360 bytes
#      ↓
# 15 KB
#
# This provides a quick confirmation that the
# PKL file has been successfully generated and
# contains the serialized chatbot package.

print("Size:", round(os.path.getsize(pkl_path) / 1024, 2), "KB")
```

```python
# ============================================
# CREATE STREAMLIT APP
# ============================================

app_code = r'''
import streamlit as st
import joblib
import os

# ============================================
# PAGE CONFIGURATION
# ============================================

st.set_page_config(
    page_title="HR Assistant",
    page_icon="🤖",
    layout="centered"
)

# ============================================
# LOAD MODEL
# ============================================

MODEL_PATH = "hr_chatbot.pkl"

if not os.path.exists(MODEL_PATH):

    st.error(
        "HR chatbot model not found. "
        "Please keep hr_chatbot.pkl in the same folder as app.py."
    )

    st.stop()

chatbot = joblib.load(MODEL_PATH)

model = chatbot["model"]
responses = chatbot["responses"]
threshold = chatbot["threshold"]


# ============================================
# CUSTOM CSS
# ============================================

st.markdown("""
<style>

.main-title {
    font-size: 38px;
    font-weight: 700;
    text-align: center;
}

.subtitle {
    text-align: center;
    font-size: 17px;
    margin-bottom: 25px;
}

.chat-user {
    background-color: #f0f2f6;
    padding: 12px;
    border-radius: 12px;
    margin: 8px 0;
}

.chat-bot {
    background-color: #e8f4ff;
    padding: 12px;
    border-radius: 12px;
    margin: 8px 0;
}

</style>
""", unsafe_allow_html=True)


# ============================================
# HEADER
# ============================================

st.markdown(
    '<div class="main-title">🤖 HR Assistant</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="subtitle">'
    'Your AI-powered Human Resources Assistant'
    '</div>',
    unsafe_allow_html=True
)


# ============================================
# SESSION STATE
# ============================================

if "messages" not in st.session_state:

    st.session_state.messages = []


# ============================================
# SIDEBAR
# ============================================

with st.sidebar:

    st.header("HR Assistant")

    st.write("I can help you with:")

    st.write("✅ Leave & Attendance")
    st.write("✅ Salary & Payroll")
    st.write("✅ Work From Home")
    st.write("✅ Employee Benefits")
    st.write("✅ Recruitment")
    st.write("✅ Interviews")
    st.write("✅ Onboarding")
    st.write("✅ Resignation")
    st.write("✅ HR Policies")

    st.divider()

    if st.button("Clear Chat"):

        st.session_state.messages = []

        st.rerun()


# ============================================
# DISPLAY CHAT HISTORY
# ============================================

for message in st.session_state.messages:

    if message["role"] == "user":

        st.markdown(
            f'''
            <div class="chat-user">
            <b>You:</b> {message["content"]}
            </div>
            ''',
            unsafe_allow_html=True
        )

    else:

        st.markdown(
            f'''
            <div class="chat-bot">
            <b>HR Assistant:</b> {message["content"]}
            </div>
            ''',
            unsafe_allow_html=True
        )


# ============================================
# USER INPUT
# ============================================

user_input = st.chat_input(
    "Ask your HR question..."
)


# ============================================
# CHATBOT LOGIC
# ============================================

if user_input:

    # Store user message
    st.session_state.messages.append({
        "role": "user",
        "content": user_input
    })

    # Predict probability
    probabilities = model.predict_proba([user_input])[0]

    max_probability = probabilities.max()

    predicted_intent = model.classes_[
        probabilities.argmax()
    ]

    # Generate response
    if max_probability < threshold:

        bot_response = (
            "I'm sorry, I couldn't confidently understand your question. "
            "Please try asking about leave, attendance, salary, payroll, "
            "benefits, recruitment, onboarding, resignation, or HR policies."
        )

    else:

        bot_response = responses[predicted_intent]

    # Store bot response
    st.session_state.messages.append({
        "role": "assistant",
        "content": bot_response
    })

    st.rerun()


# ============================================
# FOOTER
# ============================================

st.divider()

st.caption(
    "AI HR Assistant • Please verify important employment decisions "
    "with your HR department."
)
'''

with open("/content/app.py", "w", encoding="utf-8") as f:
    f.write(app_code)

print("app.py created successfully!")
```
